```python
import pandas as pd
import math
```

```python
print(df.isnull().sum())
```

```python
original = df["humidity"].copy()

missing_rows = df.index[100:110]

df.loc[missing_rows, "humidity"] = None

print("Missing values created:")
print(df["humidity"].isnull().sum())
```

```python
data = df[["temperature_celsius", "humidity"]].dropna()

x = data["temperature_celsius"].tolist()
y = data["humidity"].tolist()

n = len(x)

sx = sy = sxy = sx2 = 0

for i in range(n):
    sx += x[i]
    sy += y[i]
    sxy += x[i] * y[i]
    sx2 += x[i] * x[i]

mean_x = sx / n
mean_y = sy / n

slope = (sxy - n * mean_x * mean_y) / (sx2 - n * mean_x * mean_x)
intercept = mean_y - slope * mean_x

print("Slope =", slope)
print("Intercept =", intercept)
```

```python
for i in missing_rows:
    temp = df.loc[i, "temperature_celsius"]

    prediction = intercept + slope * temp
    df.loc[i, "humidity"] = prediction

print("Predicted humidity values:")
print(df.loc[missing_rows, ["temperature_celsius", "humidity"]])
```

```python
error = 0

for i in missing_rows:
    actual = original[i]
    predicted = df.loc[i, "humidity"]

    error += abs(actual - predicted)

mae = error / len(missing_rows)

print("Mean Absolute Error =", mae)
```
