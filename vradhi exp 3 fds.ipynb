activity 2:

```python
n = len(temperature)
```

```python
total = 0

for value in temperature:
    total = total + value

mean = total / n
print("Mean = ", mean)
```

```python
values = list(temperature)
values.sort()

if n % 2 == 1:
    median = values[n // 2]
else:
    median = (values[n // 2 - 1] + values[n // 2]) / 2

print("Median =", median)
```

```python
frequency = {}

for value in temperature:
    if value in frequency:
        frequency[value] = frequency[value] + 1
    else:
        frequency[value] = 1

highest_frequency = 0
mode = None

for value in frequency:
    if frequency[value] > highest_frequency:
        highest_frequency = frequency[value]
        mode = value

print("Mode =", mode)
print("Frequency =", highest_frequency)
```

```python
total = 0

for value in temperature:
    total = total + (value - mean) ** 2

variance = total / n
print("Variance =", variance)
```

```python
standard_deviation = variance ** 0.5
print("Standard Deviation =", standard_deviation)
```

```python
Q1_position = (n + 1) / 4
Q3_position = 3 * (n + 1) / 4

print("Q1 position =", Q1_position)
print("Q3 position =", Q3_position)
```

python indexing starts from 0 therefore subtract 1 to make index

```python
Q1_index = int(Q1_position) - 1
Q3_index = int(Q3_position) - 1

Q1 = values[Q1_index]
Q3 = values[Q3_index]

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
print("Standard Deviation =", standard_deviation)
print("Variance =", variance)
print("IQR =", IQR)
```

activity 3:

```python
bins = [-30, -20, -10, 0, 10, 20, 30, 40, 50, 60, 70, 80]

groups = pd.cut(temperature, bins=bins)

print(groups.head())
```

```python
frequency = {}

for group in groups:
    if group in frequency:
        frequency[group] = frequency[group] + 1
    else:
        frequency[group] = 1

frequency = dict(sorted(frequency.items(), key=lambda x: x[0].left))

for group in frequency:
    print(group, "=", frequency[group])
```

```python
midpoints = {}

for group in frequency:
    lower = group.left
    upper = group.right

    midpoint = (lower + upper) / 2

    midpoints[group] = midpoint

    print(group, "Midpoint =", midpoint)
```

```python
total_fx = 0
total_f = 0

for group in frequency:
    f = frequency[group]
    x = midpoints[group]

    total_fx = total_fx + (f * x)
    total_f = total_f + f

mean = total_fx / total_f

print("Grouped Mean =", mean)
```

```python
highest_frequency = 0
mode_group = None

for group in frequency:
    if frequency[group] > highest_frequency:
        highest_frequency = frequency[group]
        mode_group = group

print("Modal Group =", mode_group)
print("Frequency =", highest_frequency)
```

```python
cumulative_frequency = 0

for group in frequency:
    cumulative_frequency = cumulative_frequency + frequency[group]

    print(group, "Frequency =", frequency[group],
          "Cumulative Frequency =", cumulative_frequency)
```

```python
N = total_f

median_position = N / 2

cumulative_frequency = 0

for group in frequency:
    cumulative_frequency = cumulative_frequency + frequency[group]

    if cumulative_frequency >= median_position:
        median_group = group
        break

print("Median Position =", median_position)
print("Median Group =", median_group)
```

```python
L = median_group.left
h = median_group.right - median_group.left
f = frequency[median_group]

CF = 0

for group in frequency:
    if group == median_group:
        break

    CF = CF + frequency[group]

median = L + ((median_position - CF) / f) * h

print("Grouped Median =", median)
```

```python
Q1_position = N / 4

cumulative_frequency = 0

for group in frequency:
    cumulative_frequency = cumulative_frequency + frequency[group]

    if cumulative_frequency >= Q1_position:
        Q1_group = group
        break

print("Q1 Position =", Q1_position)
print("Q1 Group =", Q1_group)
```

```python
L = Q1_group.left
h = Q1_group.right - Q1_group.left
f = frequency[Q1_group]

CF = 0

for group in frequency:
    if group == Q1_group:
        break

    CF = CF + frequency[group]

Q1 = L + ((Q1_position - CF) / f) * h

print("Q1 =", Q1)
```

```python
Q3_position = 3 * N / 4

cumulative_frequency = 0

for group in frequency:
    cumulative_frequency = cumulative_frequency + frequency[group]

    if cumulative_frequency >= Q3_position:
        Q3_group = group
        break

print("Q3 Position =", Q3_position)
print("Q3 Group =", Q3_group)
```

```python
L = Q3_group.left
h = Q3_group.right - Q3_group.left
f = frequency[Q3_group]

CF = 0

for group in frequency:
    if group == Q3_group:
        break

    CF = CF + frequency[group]

Q3 = L + ((Q3_position - CF) / f) * h

print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1

print("IQR =", IQR)
```

```python
total = 0

for group in frequency:
    f = frequency[group]
    x = midpoints[group]

    total = total + f * (x - mean) ** 2

variance = total / total_f

print("Grouped Variance =", variance)
```

```python
standard_deviation = variance ** 0.5

print("Grouped Standard Deviation =", standard_deviation)
```

```python
print("Frequency =", frequency)
print("Mean =", mean)
print("Mode Group =", mode_group)
print("Median =", median)
print("Variance =", variance)
print("Standard Deviation =", standard_deviation)
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```
